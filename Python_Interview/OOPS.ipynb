- OOPs (Object-Oriented Programming) in Python is a programming paradigm that organizes code into reusable units called classes and objects rather than just functions and logic.
- Think of OOP like building with LEGO blocks: instead of writing one long list of sequential instructions (procedural programming), you build self-contained blocks (objects) that interact with each other to form a scalable system.

## The Two Core Building Blocks
**Class**
- A Class is an abstract blueprint or template used to create specific objects. It defines the characteristics (attributes) and capabilities (methods) that the resulting objects will possess

**Object**
- An Object is a concrete instance of a class. When a class is defined, no memory is allocated until you instantiate an object from it.

````python
class House:
    def __init__(self, color, floors): # Constructor to set up attributes
        self.color = color             # Attribute
        self.floors = floors           # Attribute

    def open_door(self):               # Method (Function inside a class)
        return "The door is now open."
    
    
# Instantiating two completely independent objects from the House blueprint
my_house = House(color="Blue", floors=2)
neighbor_house = House(color="Green", floors=1)

print(my_house.color)        # Output: Blue
print(my_house.open_door())  # Output: The door is now open.
````

#### Encapsulation
- Encapsulation is the object-oriented programming (OOP) practice of bundling data (attributes) and methods (functions) into a single unit (a class) while restricting direct access to some of the object's internal components. Think of it like a medical capsule: the implementation details are hidden inside, and you can only interact with it through a safe, public interface.
- protected keywords). Instead, Python relies on naming conventions and a mechanism called name mangling to control data access.

**Three Levels of Access Control**
1. Public Members
    - Behavior: Accessible from anywhere—inside or outside the class.
    - Convention: Default state for all attributes and methods in Python.
- Example: self.name
2. Protected Members
    - Behavior: Accessible within the class and its subclasses.
    - Convention: Prefixing the name with a single underscore (_).
    - Note: This is purely an honor-system rule for developers ("we are all consenting adults here"). Python will not stop external code from modifying it.
    - Example: self._database_id
3. Private Members
    - Behavior: Intended to be accessible only within the class itself.
    - Convention: Prefixing the name with a double underscore (__).
    - Mechanism: Python enforces this using name mangling, which automatically renames __variable to _ClassName__variable behind the scenes, making accidental external changes much harder.
    - Example: self.__balance

````python
class BankAccount:

  def __init__(self, owner, balance, pin):
    self.owner = owner  # Public member
    self._balance = balance  # Protected member
    self.__pin = pin  # Private member

  # Public method
  def get_balance(self):
    return self._balance

  # Protected method
  def _update_balance(self, amount):
    self._balance += amount

  # Private method
  def __check_pin(self, input_pin):
    return self.__pin == input_pin

  # Public method to use private/protected parts safely
  def deposit(self, amount, pin):
    if self.__check_pin(pin):
      self._update_balance(amount)
      return f"Deposited {amount}. New balance: {self._balance}"
    return "Wrong PIN!"


# Usage
account = BankAccount("Alice", 1000, 1234)

# Public access works
print(account.owner)  # Alice
print(account.get_balance())  # 1000

# Protected access works, but is discouraged outside the class
print(account._balance)  # 1000

# Private access throws an AttributeError
# print(account.__pin)

# Accessing via safe public method
print(account.deposit(500, 1234))  # Deposited 500. New balance: 1500
````

#### Abstraction
- Abstraction is the process of hiding complex implementation details and exposing only the essential features of an object. It allows you to focus on what an object does rather than how it does it, making code simpler, modular, and easier to maintain
- When you drive a car, you use the steering wheel, gas pedal, and brakes. You do not need to understand the internal combustion or electronics under the hood to operate it. The dashboard is the abstract interface that hides that complexity

**How Abstraction Works in Python**\
Python provides a built-in module called abc (Abstract Base Classes) to implement abstraction.
- Abstract Class: A blueprint class that cannot be instantiated directly. It is meant to be inherited by other classes.
- Abstract Method: A method declared inside an abstract class that has a name but no actual implementation code (pass). Subclasses must override and define these methods.

````python
from abc import ABC, abstractmethod

# 1. Define the Abstract Base Class
class PaymentProcessor(ABC):
    
    @abstractmethod
    def process_payment(self, amount: float):
        """Abstract method; hidden implementation details."""
        pass

    def print_receipt(self, amount: float):
        """Concrete method; shared logic directly inherited by subclasses."""
        print(f"Receipt printed for ${amount}")

# 2. Implement Concrete Subclasses
class CreditCardProcessor(PaymentProcessor):
    def process_payment(self, amount: float):
        print(f"Charging ${amount} to Credit Card via gateway encryption...")

class PayPalProcessor(PaymentProcessor):
    def process_payment(self, amount: float):
        print(f"Redirecting to PayPal API to transfer ${amount}...")

# --- Execution ---
# instance = PaymentProcessor() # ERROR: Cannot instantiate abstract class directly.

cc = CreditCardProcessor()
cc.process_payment(150.00) # Executes specific credit card logic
cc.print_receipt(150.00)   # Executes shared base logic
````

#### Polymorphism
Polymorphism in Python is a core Object-Oriented Programming (OOP) concept that translates to "many forms". It allows different types of objects to respond to the same function or method call in their own unique way, enabling you to write highly flexible and reusable code. 

**1. Built-in Polymorphism** \
Python inherently uses polymorphism in its default operators and functions without needing any classes. 

• Operator Polymorphism (): The addition operator changes behavior based on data types.
```python
print(5 + 3)          # Output: 8 (Arithmetic addition)
print("Py" + "thon")  # Output: Python (String concatenation)
```

• Function Polymorphism (): The built-in  function can accept strings, lists, or dictionaries, adapting its logic seamlessly to each. 
```python
print(len("Apple"))      # Output: 5
print(len([1, 2, 3, 4])) # Output: 4
```






**2. Polymorphism with Class Methods** \
Different classes can have methods with the identical name but entirely unique logic. You can loop through these objects and trigger the same method name. 

````python
class Cat:
    def sound(self):
        return "Meow"

class Dog:
    def sound(self):
        return "Woof"

# A unified loop executing polymorphic behavior
for animal in [Cat(), Dog()]:
    print(animal.sound()) 
    
# Output:
# Meow
# Woof
````

**3. Polymorphism with Inheritance (Method Overriding)** \
When a child class modifies a method inherited from a parent class, it is known as Method Overriding. This represents runtime polymorphism because Python determines which method to execute while the script is running.

````python
class Employee:
    def get_role(self):
        return "General Staff"

class Manager(Employee):
    def get_role(self):  # Overrides parent method
        return "Team Leader"

# Python dynamically resolves the correct method at runtime
staff1 = Employee()
staff2 = Manager()

print(staff1.get_role())  # Output: General Staff
print(staff2.get_role())  # Output: Team Leader
````

**4. Duck Typing** \
Python relies heavily on a concept called Duck Typing. The philosophy comes from the phrase: "If it walks like a duck and quacks like a duck, it's a duck." [1, 2, 8, 9]  
In Python, the explicit class type of an object does not matter. What matters is whether the object has the required method or behavior.

````python
class FlyableSpacetimeMachine:
    def fly(self):
        print("Warping through space!")

class Airplane:
    def fly(self):
        print("Flying through clouds!")

# This function doesn't care about the class type, only that it can .fly()
def launch_vehicle(vehicle_obj):
    vehicle_obj.fly()

launch_vehicle(FlyableSpacetimeMachine())  # Output: Warping through space!
launch_vehicle(Airplane())                # Output: Flying through clouds!
````

**Summary of Benefits** \
- Code Reusability: Write one interface or function that can handle many diverse object types. 
- Maintainability: Easily add new classes without altering existing functions or loops. 
- Clean Code: Eliminates the need for long, messy conditional structures () checking for data types

#### Inheritance
- Inheritance is a core concept in Object-Oriented Programming (OOP) that allows a new class (called a child or derived class) to adopt all the attributes and methods of an existing class (called a parent or base class). This mechanism promotes code reusability by preventing redundant code rewriting.
- To inherit from a class, pass the parent class name inside parentheses when defining the child class

````python
# Parent Class
class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        return "Some generic animal sound"

# Child Class inheriting from Animal
class Dog(Animal):
    def __init__(self, name, breed):
        # super() links to Animal, initializing the 'name' attribute
        super().__init__(name) 
        self.breed = breed

    def make_sound(self):
        # You can also use super() to call parent regular methods
        parent_sound = super().make_sound()
        return f"{parent_sound}, but specifically a loud Bark!"

# Create an instance of the child class
my_dog = Dog(name="Buddy", breed="Golden Retriever")

print(f"Name: {my_dog.name}")
print(f"Breed: {my_dog.breed}")
print(f"Sound: {my_dog.make_sound()}")
````

**⚙️ Core Concepts** 
- Method Overriding: A child class can provide a specific implementation for a method that is already defined in its parent class to customize behavior.
- The super() Function: Used inside a child class to call methods from the parent class, which is highly useful when extending the parent's __init__ constructor.
- The isinstance() Function: Returns True if an object is an instance of a specific class or any of its parent classes.

**🔀 Types of Inheritance in Python** 
- Single Inheritance: A child class inherits from exactly one parent class.
- Multiple Inheritance: A child class inherits directly from multiple parent classes simultaneously.
- Multilevel Inheritance: A child class inherits from a parent class, which in turn inherits from a grandparent class.
- Hierarchical Inheritance: Multiple child classes inherit from the same single parent class.
