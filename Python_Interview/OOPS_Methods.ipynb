#### Static and Class Methods
In Python, class methods and static methods are specialized types of methods bound directly to the class rather than its individual instances.
- Class methods receive the class itself as an implicit first parameter (cls) and can access or modify class-level state.
- Static methods do not receive any implicit first parameter (neither self nor cls) and behave exactly like plain functions grouped inside the class namespace for organizational purposes.

```python
from datetime import date


class Employee:
    # Class-level attribute
    company_name = "Tech Corp"

    def __init__(self, name: str, age: int):
        # Instance attributes
        self.name = name
        self.age = age

    # 1. Instance Method
    def display_details(self):
        # Accesses individual object state via 'self'
        print(f"Name: {self.name}, Age: {self.age}, Company: {self.company_name}")

    # 2. Class Method
    @classmethod
    def from_birth_year(cls, name: str, birth_year: int):
        # Accesses class reference 'cls' to act as an alternative constructor
        calculated_age = date.today().year - birth_year
        # Returns a new instance of the class (equivalent to Employee(name, calculated_age))
        return cls(name, calculated_age)

    # 3. Static Method
    @staticmethod
    def is_adult(age: int) -> bool:
        # Isolated function; does not accept 'self' or 'cls'
        return age >= 18


# --- Usage Demonstration ---

# Utilizing the Class Method as a factory to create an instance
emp1 = Employee.from_birth_year("Alice", 1995)
emp1.display_details()  # Output: Name: Alice, Age: 31, Company: Tech Corp

# Utilizing the Static Method independently without instantiating the class
status = Employee.is_adult(22)
print(f"Is adult? {status}")  # Output: Is adult? True
```

#### Factory Methods
In Python, a factory method is a creational design pattern that provides an interface for creating objects but allows subclasses or specialized methods to alter the type of objects that will be created. Instead of calling a class constructor directly (e.g., Car()), the client code calls a factory method, which centralizes and encapsulates the instantiation logic.

```python
import json

class User:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    # Factory method 1: Create an instance from a JSON string
    @classmethod
    def from_json(cls, json_string: str):
        data = json.loads(json_string)
        return cls(data["name"], data["age"])

    # Factory method 2: Create an instance with default/guest values
    @classmethod
    def create_guest(cls):
        return cls(name="Guest User", age=0)

# Client Usage
user1 = User("Alice", 30)                  # Standard instantiation
user2 = User.from_json('{"name": "Bob", "age": 25}') # Factory instantiation
user3 = User.create_guest()               # Factory instantiation
```

#### Dunder Methods
Dunder methods (short for Double Underscore methods) are special, built-in methods in Python surrounded by two underscores, such as __init__ and __str__. Also known as magic methods, they allow you to hook into Python's core functionality, enabling operator overloading and mimicking the behavior of native Python types within your custom classes

Core Categories of Dunder MethodsPython groups its magic methods into distinct functional areas based on the official Python Data Model:
1. Object Initialization & Representation\
    These manage how an object is created and how it displays itself to users and developers.
    - __init__(self, ...) – The initializer method called automatically when creating an object.
    - __str__(self) – Returns a user-friendly, human-readable string representation (triggered by print() or str()).
    - __repr__(self) – Returns an unambiguous string representation for developers and debugging (triggered by repr()).
2. Mathematical Operations & Operator 
    OverloadingThese let you define how operators like +, -, or * handle your custom objects.
    - __add__(self, other) – Overloads the addition operator (+).
    - __sub__(self, other) – Overloads the subtraction operator (-).
    - __mul__(self, other) – Overloads the multiplication operator (*).
    - __truediv__(self, other) – Overloads the standard division operator (/).
3. Comparison Operators\
    These specify how your custom data structures are ordered or compared.
    - __eq__(self, other) – Handles equality checks (==).
    - __lt__(self, other) – Handles "less than" checks (<).
    - __le__(self, other) – Handles "less than or equal to" checks (<=).
4. Container & Sequence Emulation\
    These make your class behave like built-in lists, dictionaries, or sets.
    - __len__(self) – Returns the size of the object when passed to len().
    - __getitem__(self, key) – Enables square-bracket indexing and slicing, like obj[key].
    - __iter__(self) – Returns an iterator object, enabling for loops.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Changes how the vector prints
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    # Overloads the + operator
    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    # Overloads the == operator
    def __eq__(self, other):
        if isinstance(other, Vector):
            return self.x == other.x and self.y == other.y
        return False

# Usage
v1 = Vector(2, 4)
v2 = Vector(1, 3)

print(v1)          # Outputs: Vector(2, 4)  (Triggers __str__)
print(v1 + v2)     # Outputs: Vector(3, 7)  (Triggers __add__)
print(v1 == v2)    # Outputs: False         (Triggers __eq__)
```

#### Recursion Method
Recursion in Python is a programming technique where a function calls itself to solve a problem by breaking it down into smaller, manageable subproblems. Every recursive method requires two essential components: a base case to stop the execution and a recursive case to move toward the stopping condition

```python
def factorial(n):
    if n <= 1:  # Base case
        return 1
    return n * factorial(n - 1)  # Recursive case

print(factorial(5))  # Output: 120
```
